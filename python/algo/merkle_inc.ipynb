```python
# 증분 머클 트리

ZERO = 0
# 0 = 리프 노드
HEIGHT = 3

def hash(a, b):
    return f'h({a},{b})'

class MerkleTree:
    def __init__(self):
        # 모든 리프 = ZERO인 트리 계산
        zeros = [ZERO] * HEIGHT
        for i in range(HEIGHT - 1):
            zeros[i + 1] = hash(zeros[i], zeros[i])

        # 각 레벨의 제로 해시
        self.zeros = zeros
        # 각 레벨에서 모든 자손이 채워진 왼쪽 노드 중 가장 오른쪽 노드
        self.nodes = [ZERO] * HEIGHT
        # 다음 리프 인덱스
        self.next = 0
        # 리프
        self.vals = [ZERO] * (2**HEIGHT)

    def root(self):
        h = self.zeros[0]
        k = self.next
        
        for i in range(HEIGHT):
            if k % 2 == 1:
                h = hash(self.nodes[i], h)
            else:
                h = hash(h, self.zeros[i])
            k //= 2

        return h

    def insert(self, val):
        assert self.next < 2 ** HEIGHT - 1

        self.vals[self.next] = val
        self.next += 1

        h = val
        k = self.next

        for i in range(HEIGHT):
            if k % 2 == 1:
                self.nodes[i] = h
                return
            else:
                h = hash(self.nodes[i], h)
            k //= 2
```

```python
tree = MerkleTree()

print(0, tree.root())

for i in range(1, 2**HEIGHT):
    tree.insert(i)
    print(i, tree.root())

print(tree.vals)
```
